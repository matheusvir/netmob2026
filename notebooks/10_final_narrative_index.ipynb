# Guia da Análise Final — NetMob 2026



## 1. Visão geral do projeto

Este projeto investiga como as condições meteorológicas afetam a demanda de passageiros e o desempenho operacional do sistema de ônibus de Niterói/RJ durante março de 2026. A análise integra dados meteorológicos, registros de bilhetagem, telemetria de mobilidade e contexto GTFS para estudar tanto os impactos sobre a demanda quanto sobre a operação do serviço.


## 2. Perguntas de pesquisa

A análise final está organizada em torno de cinco perguntas:

1. Há variação significativa no volume de passageiros associada a condições climáticas adversas em comparação com dias de tempo firme?
2. A resposta da demanda ao clima difere entre perfis de passageiros?
3. Durante condições climáticas adversas, há aumento perceptível no intervalo entre ônibus ou redução da velocidade média operacional?
4. As rotas de ônibus apresentam níveis distintos de resiliência às condições climáticas, com algumas linhas mais vulneráveis que outras?
5. A redução de frequência e velocidade durante episódios de chuva intensifica a queda da demanda, ou a demanda cai de forma independente da qualidade do serviço naquele momento?


## 3. Organização do repositório

O repositório está estruturado como um fluxo analítico centrado em notebooks:

- `00_build_preprocessed_datasets.ipynb`: constrói os datasets derivados necessários para as análises finais.
- `01_eda_weather.ipynb`: EDA de clima e base de pré-processamento meteorológico.
- `02_eda_ticket.ipynb`: EDA de demanda via bilhetagem e base de pré-processamento.
- `03_eda_mobility.ipynb`: EDA de mobilidade e base de pré-processamento operacional.
- `04_build_analytic_base.ipynb`: valida e audita a base analítica integrada.
- `05_analysis_demand_weather.ipynb`: analisa a demanda de passageiros sob variação climática.

- `07_analysis_route_resilience.ipynb`: compara a resiliência entre rotas.
- `08_analysis_joint_demand_service.ipynb`: examina conjuntamente os efeitos sobre demanda e serviço.
- `09_storytelling_outputs.ipynb`: organiza as saídas finais para síntese e apresentação.

## 6. Resumo dos principais achados

Os notebooks finais executados mostram um quadro geral consistente: março de 2026 contém majoritariamente dias de tempo firme, um número menor de dias com chuva leve/moderada e nenhum dia inteiro classificado como **Heavy Rain / Storm** na categorização diária final. Isso significa que as conclusões mais fortes dizem respeito ao contraste entre tempo firme e chuva leve/moderada, enquanto afirmações sobre clima extremo precisam ser feitas com cautela.

### Q1 — Variação da demanda sob clima adverso
Os modelos de demanda indicam uma associação negativa entre chuva e embarques de passageiros, mas o efeito é modesto e apenas limítrofe em significância na especificação mais ampla em nível rota-hora. No modelo de Poisson por rota-hora, o efeito estimado é de aproximadamente **-1,71% de embarques por milímetro adicional de chuva**, enquanto a especificação diária produz um coeficiente negativo pequeno, com intervalo de confiança próximo de zero. As evidências detalhadas estão em `05_analysis_demand_weather.ipynb`.

### Q2 — Diferenças entre perfis de passageiros
A resposta ao clima é heterogênea entre os grupos de passageiros, especialmente sob chuva moderada. O notebook executado mostra quedas relativas mais fortes para embarques de **Student** e **Senior** sob chuva moderada, enquanto usuários **Cash** apresentam o coeficiente negativo de chuva mais claramente estimado no modelo com interações. As evidências detalhadas estão em `05_analysis_demand_weather.ipynb`.

### Q3 — Degradação operacional sob clima adverso
Os indicadores operacionais sugerem que a chuva está associada de forma mais clara à **redução de velocidade** do que ao aumento sistemático de headway. Nos modelos de sensibilidade executados, o coeficiente de chuva para `speed_p50` é consistentemente negativo e estatisticamente significativo, enquanto o coeficiente de chuva para `headway_p50` é positivo, mas não significativo. A amostra limpa integrada cobre **19 dias**, com **4 dias de mobilidade marcados como parciais** e testados separadamente. As evidências detalhadas estão em `06_analysis_service_vs_weather_gtfs.ipynb`.

### Q4 — Diferenças de resiliência entre rotas
As rotas não reagem de maneira uniforme às condições de chuva. O notebook de resiliência identifica variação relevante entre linhas, com **41 rotas** mantendo cobertura tanto em dias de tempo firme quanto em dias de chuva no painel final de comparação. As rotas mais vulneráveis no ranking final incluem **34B, 37, 54, 15 e 45**, indicando que a resiliência em nível de rota é heterogênea, e não uniforme em todo o sistema. As evidências detalhadas estão em `07_analysis_route_resilience.ipynb`.

### Q5 — Efeitos conjuntos entre demanda e serviço
Quando controles operacionais são adicionados ao modelo de demanda, o efeito direto estimado da chuva permanece negativo, mas os termos de interação entre chuva e métricas de serviço não apresentam força estatística no comparativo final entre modelos aninhados. Isso sugere que a chuva está associada a menor demanda, mas que a evidência de que a degradação operacional amplifica de forma relevante essa perda de demanda é limitada nesta amostra. As evidências detalhadas estão em `08_analysis_joint_demand_service.ipynb` e nos visuais consolidados em `09_storytelling_outputs.ipynb`.
